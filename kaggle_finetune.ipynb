```python
import os, torch
from datasets import load_dataset
from transformers import (
    AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig,
    Trainer, TrainingArguments, DataCollatorForLanguageModeling,
)
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training

MODEL_ID   = 'Qwen/Qwen2.5-3B-Instruct'
DATASET    = '/kaggle/working/security_audit_dataset.jsonl'
OUTPUT_DIR = '/kaggle/working/security_specialist_lora'

# --- Tokenizer ---
print('Loading tokenizer...', flush=True)
tokenizer = AutoTokenizer.from_pretrained(MODEL_ID, trust_remote_code=True)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = 'right'
print(f'Vocab size: {len(tokenizer)}', flush=True)

# --- System prompt ---
with open('/kaggle/working/tune.md') as f:
    system_prompt = f.read().strip()
print(f'System prompt: {len(system_prompt)} chars', flush=True)

# --- Dataset ---
print('Loading dataset...', flush=True)
dataset = load_dataset('json', data_files=DATASET, split='train')
print(f'{len(dataset)} examples', flush=True)

def format_prompts(ex):
    msgs = [{'role': 'system', 'content': system_prompt}]
    msgs += [m for m in ex['messages'] if m['role'] != 'system']
    return {'text': tokenizer.apply_chat_template(msgs, tokenize=False, add_generation_prompt=False)}

formatted = dataset.map(format_prompts)
print(f'Sample length: {len(formatted[0]["text"])} chars', flush=True)

def tokenize_fn(ex):
    out = tokenizer(ex['text'], truncation=True, max_length=4096, padding=False)
    out['labels'] = out['input_ids'].copy()
    return out

tokenized = formatted.map(tokenize_fn, remove_columns=['text'])
data_collator = DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm=False)

# --- Model (4-bit QLoRA) ---
print('Loading model (4-bit)...', flush=True)
bnb = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type='nf4',
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True,
)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID, quantization_config=bnb, device_map='auto',
    torch_dtype=torch.float16, trust_remote_code=True,
)
model.config.use_cache = False
model = prepare_model_for_kbit_training(model)
print(f'Model loaded. VRAM: {torch.cuda.memory_allocated(0)/1e9:.2f} GB', flush=True)

# --- LoRA ---
lora_cfg = LoraConfig(
    r=16, lora_alpha=32, lora_dropout=0.05, bias='none', task_type='CAUSAL_LM',
    target_modules=['q_proj','k_proj','v_proj','o_proj','gate_proj','up_proj','down_proj'],
)
model = get_peft_model(model, lora_cfg)
model.print_trainable_parameters()

# --- Train ---
args = TrainingArguments(
    output_dir='/kaggle/working/checkpoints',
    num_train_epochs=3,
    per_device_train_batch_size=1,
    gradient_accumulation_steps=4,
    learning_rate=2e-4,
    lr_scheduler_type='cosine',
    warmup_steps=1,
    logging_steps=1,
    fp16=True,
    optim='paged_adamw_8bit',
    gradient_checkpointing=True,
    save_strategy='no',
    report_to='none',
)
trainer = Trainer(model=model, args=args, train_dataset=tokenized, data_collator=data_collator)

print(f'Starting training... VRAM: {torch.cuda.memory_allocated(0)/1e9:.2f} GB', flush=True)
result = trainer.train()
print(f'Done. Loss: {result.training_loss:.4f}  Peak VRAM: {torch.cuda.max_memory_allocated(0)/1e9:.2f} GB', flush=True)

# --- Save ---
print(f'Saving to {OUTPUT_DIR}...', flush=True)
trainer.model.save_pretrained(OUTPUT_DIR)
tokenizer.save_pretrained(OUTPUT_DIR)
print('Saved!', flush=True)
```

```python
import subprocess, sys, torch
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', '-U', 'torchao>=0.16.0'], check=False)

from peft import PeftModel
from transformers import AutoModelForCausalLM, AutoTokenizer

BASE = 'Qwen/Qwen2.5-3B-Instruct'
LORA = '/kaggle/working/security_specialist_lora'

print('Loading fine-tuned model for inference...', flush=True)
tok  = AutoTokenizer.from_pretrained(LORA)
base = AutoModelForCausalLM.from_pretrained(BASE, torch_dtype=torch.float16, device_map='auto', trust_remote_code=True)
mdl  = PeftModel.from_pretrained(base, LORA)
mdl.eval()
print('Model ready.', flush=True)

with open('/kaggle/working/tune.md') as f:
    system_prompt = f.read().strip()

question = "How do I safely test for SQL injection in an authorized web application pentest? Include safe payloads and what to look for."

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user",   "content": question},
]
prompt = tok.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
inputs = tok(prompt, return_tensors='pt').to(mdl.device)

with torch.no_grad():
    out = mdl.generate(**inputs, max_new_tokens=512, temperature=0.1, do_sample=True, pad_token_id=tok.eos_token_id)

response = tok.decode(out[0][inputs.input_ids.shape[1]:], skip_special_tokens=True)
print("\n=== Model Response ===")
print(response)
```
